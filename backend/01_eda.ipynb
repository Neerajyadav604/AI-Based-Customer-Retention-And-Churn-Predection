```python
import pandas as pd

# The file is in Latin-1 encoding (it contains the £ sign), so say so
df = pd.read_csv("data/online_retail_II.csv", encoding="latin-1")

print(df.shape)              # rows, columns
print(df.columns.tolist())   # column names
df.head()                    # first 5 rows
```

```python
df.info()                  # data types and non-empty counts
print(df.isnull().sum())   # missing values per column
df.describe()              # basic statistics for number columns
```

```python
# Step 1: convert the date text into real dates
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
print("Dates:", df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())

# Step 2: drop rows with no customer ID
df = df.dropna(subset=["Customer ID"])

# Step 3: drop cancelled invoices (invoice number starts with "C")
df = df[~df["Invoice"].astype(str).str.startswith("C")]

# Step 4: keep only rows with positive quantity and price
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)]

# Step 5: customer ID as a whole number, and amount per row
df["Customer ID"] = df["Customer ID"].astype(int)
df["Amount"] = df["Quantity"] * df["Price"]

print("Rows left:", df.shape[0])
print("Unique customers:", df["Customer ID"].nunique())

# Save the cleaned data for the next notebook
df.to_csv("data/clean_orders.csv", index=False)
```
